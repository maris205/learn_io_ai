## 文本分类Agent

您简单介绍一下“文本分类”。

我们可以把它想象成一个**拥有超能力的图书管理员**。

当一大堆杂乱无章的书被送到图书馆时，这位图书管理员能以极快的速度浏览每一本书的内容，然后准确地把它放到对应的书架上——“历史类”、“科学类”、“小说类”等等。

---

### 什么是文本分类？

**文本分类**（Text Classification），是自然语言处理（NLP）中最基础、也最重要的任务之一。它的目标就是让计算机**自动地给一段文本内容分配一个或多个预先定义好的“类别”或“标签”**。

简单来说，就是让机器“读懂”一段话，然后告诉我们“这段话是关于什么的？”

### 一个简单的例子

假设我们想让计算机自动分辨电影评论的情感倾向，我们可以预先定义三个类别：`正面`、`负面`、`中性`。

* **输入文本 1:**
  > “这部电影太棒了，演员演技在线，情节也很吸引人！”
* **计算机的输出标签:**
  > `正面`

* **输入文本 2:**
  > “完全是浪费时间，故事毫无逻辑，不推荐观看。”
* **计算机的输出标签:**
  > `负面`

这个过程就是一次典型的文本分类。



### 计算机是如何“学习”分类的？

计算机并不是天生就会分类，它需要一个“学习”的过程，通常分为两步：

1.  **训练 (Training)**：
    我们先准备成千上万篇已经由人工分好类的文章（这就像是给计算机一本“标准答案练习册”）。然后，我们让一个AI模型（比如现在流行的大语言模型）去“学习”这些数据。模型会自己寻找规律，比如看到“棒极了”、“推荐”这些词的文章，大概率是`正面`的；而看到“失望”、“浪费时间”这些词的文章，则很可能是`负面`的。

2.  **预测 (Prediction / Inference)**：
    当模型“学成”之后，我们就可以给它一篇它从未见过的、没有标签的新文章。它会利用在训练中学到的知识和规律，来预测这篇文章最应该属于哪个类别。

### 为什么文本分类很重要？（主要应用）

文本分类技术几乎无处不在，是许多智能应用的核心：

* **垃圾邮件过滤 (Spam Detection)**
    * 这是最经典的文本分类应用。你的邮箱会自动将收到的邮件分为“收件箱”或“垃圾邮件箱”。

* **情感分析 (Sentiment Analysis)**
    * 分析社交媒体（如Twitter、微博）上的帖子或电商网站上的商品评论，判断用户对某个产品、品牌或事件的情感是正面的、负面的还是中性的。

* **新闻内容分类 (News Categorization)**
    * 像今日头条、Google News这样的新闻App，会自动将新闻文章分为“体育”、“科技”、“财经”、“娱乐”等不同频道。

* **智能客服与工单系统 (Customer Support)**
    * 自动将用户发来的支持请求或投诉邮件，根据内容分类给正确的处理部门（例如“技术问题”、“账单问题”、“物流咨询”）。

* **内容审核 (Content Moderation)**
    * 自动识别网络上的文本内容是否包含有害信息，如仇恨言论、暴力或垃圾广告，并进行处理。

**总而言之，文本分类的核心价值就是将海量的、非结构化的文本信息，变得井井有条、结构清晰**，从而让计算机能够更高效地处理和利用这些信息。它是实现信息自动化的关键一步。

## 命名实体识别Agent


我们可以把它想象成让电脑**用不同颜色的荧光笔，自动在文章中划出重点**。

---

### 什么是命名实体识别？

**命名实体识别**（Named Entity Recognition, 简称 **NER**），是自然语言处理（NLP）中的一项基础核心任务。它的目标就是从一段文本中，**找出并分类预先定义好的“实体”**。

简单来说，就是在一句话里找出“**谁 (Who)**”、“**什么 (What)**”、“**何时 (When)**”、“**何地 (Where)**”这些关键信息。

### 一个简单的例子

来看下面这句话：

> **苹果公司**的首席执行官**蒂姆·库克**于**2024年9月10日**在**加州**的**库比蒂诺**发布了新款iPhone。

如果让计算机来做命名实体识别，它会像这样“划重点”：

* `苹果公司` → **组织机构 (Organization)**
* `蒂姆·库克` → **人名 (Person)**
* `2024年9月10日` → **日期 (Date)**
* `加州` → **地理位置 (Location)**
* `库比蒂诺` → **地理位置 (Location)**



计算机不仅找到了这些词，还准确地给它们贴上了预先定义好的标签（例如：人名、地名等）。

### 常见的实体类别有哪些？

NER可以识别的实体类型有很多，最常见的包括：

* **PER**: 人名 (Person)
* **ORG**: 组织机构名 (Organization)，如公司、政府部门
* **LOC**: 地理位置名 (Location)，如国家、城市、街道
* **GPE**: 地缘政治实体 (Geo-Political Entity)，特指国家、城市等行政区划
* **TIME**: 时间
* **DATE**: 日期
* **MONEY**: 货币
* **PRODUCT**: 产品名称
* **EVENT**: 事件名称

### 为什么命名实体识别很重要？（主要应用）

NER是许多更高级AI应用的基础，它能将非结构化的文本变得结构化，应用非常广泛：

1.  **信息提取与智能搜索**
    * 当你问搜索引擎“**百度的CEO是谁？**”时，NER首先识别出“百度”是一个**组织机构**，然后系统就可以在知识库中查找与“百度”相关的“CEO”信息。

2.  **内容推荐与分类**
    * 新闻网站通过NER识别一篇文章中提到的人物、地点和公司，从而判断这篇文章是关于科技、政治还是体育，并为你推荐类似内容。

3.  **智能客服**
    * 自动从用户的投诉邮件中提取**产品名称**、**订单号**、**日期**和**金额**等实体，快速创建工单，提高服务效率。

4.  **简历解析**
    * 自动从求职者的简历中抽取出**姓名**、**毕业院校**、**曾任职公司**和**专业技能**等信息，存入人才数据库。

5.  **舆情分析**
    * 分析社交媒体上关于某个**品牌**或**产品**的讨论，快速了解大众的情感倾向和关注点。

**总而言之，命名实体识别是让计算机读懂文本中关键信息的第一步**，是连接人类语言和结构化数据的关键桥梁。

当然，大模型时代，命名实体识别，也不需要复杂的训练流程。只需要提示词即可。

## 自定义Agent

通过提示词自定义Agent，提示词可以自定义各种任务，包括之前的各类任务

```python
from iointel import (
    Agent,
    Workflow
)

import os
import asyncio

api_key= "io-v2-eyJhbGciOiJSUzI1NiIsInR5cCI6IkpXVCJ9.eyJvd25lciI6IjAzNzRhMjJjLTVmZTctNDBhNS04OGU4LTEwMjk3MGViMjEyNiIsImV4cCI6NDkwNTc2MTYyM30.F2HzmnMxRaHpRh2kfPRSgp7JHqtVb5BUmZD4kz1TxY28Usvyh18shHpNpvmncoE8S-IvZH3qlEEScNtO9j-5Dw"

text = """The global electric vehicle (EV) market is witnessing exponential growth, driven by advancements in battery technology and government incentives. Major automakers, including Tesla, Ford, and Volkswagen, are investing heavily in EV production. However, challenges such as charging infrastructure and raw material shortages for lithium-ion batteries remain key concerns. Industry experts predict that by 2030, EVs will account for over 50% of new car sales worldwide, marking a significant shift in the automotive landscape."""

agent = Agent(
    name="Super Agent",
    instructions="You are an assistant specialized in doing anything.",
    model="meta-llama/Llama-3.3-70B-Instruct",
    api_key=api_key,
    base_url="https://api.intelligence.io.solutions/api/v1"
)

workflow = Workflow(objective=text, client_mode=False)

async def run_workflow():
    results = (await workflow.custom(name="custom-task", objective="continue summarizing",instructions="Focus on previous history",agents=[agent],).run_tasks())["results"]
    return results

results = await run_workflow()
print(results)
```
