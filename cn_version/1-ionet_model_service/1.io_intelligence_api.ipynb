## 基础大语言模型调用

ionet大模型的服务调用和openai的sdk完全兼容,具体要点如下：


### 1\. 安装 OpenAI Python 库

首先，你需要确保你的 Python 环境已经安装了 OpenAI 官方提供的库。可以通过 `pip` 这个包管理器来安装：

```bash
pip install --upgrade openai
```

加上 `--upgrade` 参数可以确保你安装的是最新版本，这很重要，因为 OpenAI 的 API 和库会不时更新。

### 2\. 获取并设置 API 密钥

要使用 OpenAI 的服务，你必须拥有一个 API 密钥（API Key），第三服务一般也有个key。

  * **获取密钥**：登录你的 OpenAI 账户，在 “API keys” 页面创建一个新的密钥。请注意，这个密钥只会完整显示一次，请务必妥善保管，不要泄露给他人。

  * **设置密钥**：在代码中使用密钥最推荐、最安全的方式是将其设置为环境变量。

      * 在 **Linux 或 macOS** 系统中，可以在终端执行：
        ```bash
        export OPENAI_API_KEY='你的密钥'
        ```
      * 在 **Windows** 系统中，可以执行：
        ```bash
        set OPENAI_API_KEY=你的密钥
        ```
        为了让这个设置永久生效，你需要将其添加到系统的环境变量配置中。

    当你在代码中初始化 OpenAI 客户端时，它会自动读取名为 `OPENAI_API_KEY` 的环境变量，无需在代码中硬编码密钥，这样更安全。

    当然，大部分兼容openai sdk的大模型服务，一般还需要设置一个base_url, 表示服务的具体服务器地址

### 3\. 初始化客户端

在你的 Python 代码中，第一步是导入 `openai` 库并创建一个客户端实例。

```python
from openai import OpenAI

# 如果你已经设置了环境变量，客户端会自动读取密钥
client = OpenAI()

# 如果你没有设置环境变量（不推荐），也可以在代码中直接指定
# client = OpenAI(api_key="你的密钥")
```

### 4\. 调用大语言模型（核心步骤）

最常见的用法是使用“聊天补全”（Chat Completions）接口。这个接口非常灵活，不仅可以用于多轮对话，也可以用于单次的文本生成任务。

下面是一个基础的调用示例，向模型提一个问题：

```python
import os
from openai import OpenAI

# 初始化客户端（会自动读取环境变量 OPENAI_API_KEY）
client = OpenAI()

try:
    # 发起 API 请求
    response = client.chat.completions.create(
        model="gpt-4o",  # 指定要使用的模型，例如 gpt-4o, gpt-4-turbo, gpt-3.5-turbo
        messages=[
            {"role": "system", "content": "你是一个乐于助人的AI助手。"},  # 系统消息，设定AI的角色和行为
            {"role": "user", "content": "请简单介绍一下新加坡环球影城有什么好玩的？"} # 用户的提问
        ]
    )

    # 提取并打印模型的回复
    ai_response = response.choices[0].message.content
    print(ai_response)

except Exception as e:
    print(f"调用 API 时出错: {e}")

```

#### 代码关键点解析：

  * **`model`**: 指定你要使用的模型。`gpt-4o` 是目前最新、功能最全面的模型之一，性价比很高。你也可以选择其他模型。
  * **`messages`**: 这是请求的核心内容，它是一个列表，其中包含一个或多个字典。
      * **`role`**: 角色，主要有三种：
          * `system`: 系统角色，用于给模型设定整体的行为指令和背景。例如，“你是一个专业的翻译家”。
          * `user`: 用户角色，代表终端用户提出的问题或指令。
          * `assistant`: 助手角色，代表模型之前的回复。在多轮对话中，你需要把历史对话（用户提问和AI回答）都放进 `messages` 列表，这样模型才能理解上下文。
      * **`content`**: 消息的具体内容。
  * **`response`**: API 返回的是一个复杂的对象。我们通常最关心的是 `response.choices[0].message.content`，这里包含了模型生成的文本回复。

### 5\. 进阶用法：控制输出

你还可以在 `create()` 方法中加入更多参数来精细地控制模型的输出：

  * **`temperature`** (温度, 0 到 2之间，默认 1): 控制输出的随机性。值越高（如 1.2），输出越随机、越有创意；值越低（如 0.2），输出越确定、越保守。
  * **`max_tokens`** (最大令牌数): 设置模型生成内容的最大长度。这有助于控制成本和响应时间。注意，一个单词通常由多个令牌（token）组成。
  * **`stream`** (流式传输): 设置为 `True` 后，API 会像打字机一样逐字返回结果，而不是等全部生成完再一次性返回。这对于实现实时聊天效果非常有用。

#### 流式传输（Streaming）示例：

```python
# ...（前面的代码相同）...

response_stream = client.chat.completions.create(
    model="gpt-4o",
    messages=[
        {"role": "user", "content": "给我写一个关于太空旅行的短故事"}
    ],
    stream=True  # 开启流式传输
)

print("AI 正在生成：")
for chunk in response_stream:
    # 检查是否有内容，并打印出来
    if chunk.choices[0].delta.content is not None:
        print(chunk.choices[0].delta.content, end="", flush=True)

print() # 换行
```
### 总结

使用 OpenAI SDK 的基本流程就是：**安装 -\> 获取密钥 -\> 初始化客户端 -\> 构造 `messages` 列表 -\> 调用 `chat.completions.create` 方法 -\> 解析返回结果**。

### 多轮回话

````python
import openai

# --- 初始化客户端 ---
# 请将 api_key 替换为您自己的密钥
client = openai.OpenAI(
    api_key="io-v2-eyJhbGciOiJSUzI1NiIsInR5cCI6IkpXVCJ9.eyJvd25lciI6IjAzNzRhMjJjLTVmZTctNDBhNS04OGU4LTEwMjk3MGViMjEyNiIsImV4cCI6NDkwNTc2MTYyM30.F2HzmnMxRaHpRh2kfPRSgp7JHqtVb5BUmZD4kz1TxY28Usvyh18shHpNpvmncoE8S-IvZH3qlEEScNtO9j-5Dw",
    base_url="https://api.intelligence.io.solutions/api/v1/",
)

# --- 对话历史记录 ---
# 首先放入"system"消息，为AI设定一个角色。
conversation_history = [
    {"role": "system", "content": "你是一个乐于助人的AI助手。"},
]

print("--- 开始一个两轮对话示例 ---")

# --- 第一轮对话 ---
# 1. 定义第一个用户问题
first_user_message = "你好，简单介绍一下你自己。"
print(f"你: {first_user_message}")

# 2. 将用户问题添加到历史记录
conversation_history.append({"role": "user", "content": first_user_message})

# 3. 发送请求
first_response = client.chat.completions.create(
    model="meta-llama/Llama-3.3-70B-Instruct",
    messages=conversation_history,
    temperature=0.7,
    stream=False,
    max_completion_tokens=150
)

# 4. 获取并打印AI的回复
first_assistant_message = first_response.choices[0].message.content
print(f"助手: {first_assistant_message}\n")

# 5. 将AI的回复也添加到历史记录，为下一轮做准备
conversation_history.append({"role": "assistant", "content": first_assistant_message})


# --- 第二轮对话 ---
# 1. 定义第二个用户问题 (这个问题是基于上一轮对话的)
second_user_message = "很好。那你能帮我做什么呢？"
print(f"你: {second_user_message}")

# 2. 将第二个问题添加到历史记录
conversation_history.append({"role": "user", "content": second_user_message})

# 3. 再次发送包含所有历史的请求
second_response = client.chat.completions.create(
    model="meta-llama/Llama-3.3-70B-Instruct",
    messages=conversation_history, # 关键点：这里传递的是包含第一轮对话的完整历史
    temperature=0.7,
    stream=False,
    max_completion_tokens=150
)

# 4. 获取并打印最终的回复
second_assistant_message = second_response.choices[0].message.content
print(f"助手: {second_assistant_message}")

````

````python
import openai

# --- 初始化客户端 ---
# 请将 api_key 替换为您自己的密钥
client = openai.OpenAI(
    api_key="io-v2-eyJhbGciOiJSUzI1NiIsInR5cCI6IkpXVCJ9.eyJvd25lciI6IjAzNzRhMjJjLTVmZTctNDBhNS04OGU4LTEwMjk3MGViMjEyNiIsImV4cCI6NDkwNTc2MTYyM30.F2HzmnMxRaHpRh2kfPRSgp7JHqtVb5BUmZD4kz1TxY28Usvyh18shHpNpvmncoE8S-IvZH3qlEEScNtO9j-5Dw",
    base_url="https://api.intelligence.io.solutions/api/v1/",
)

# --- 发起流式请求 ---
# 关键点 1: 将 stream 设置为 True
response_stream = client.chat.completions.create(
    model="meta-llama/Llama-3.3-70B-Instruct",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        # 为了让流式效果更明显，换一个需要更长回答的问题
        {"role": "user", "content": "Tell me a short story about a robot who dreams of stars."},
    ],
    temperature=0.7,
    stream=True,  # 设置为 True 来开启流式传输
    max_completion_tokens=150
)

# --- 处理并打印数据流 ---
print("助手: ", end="")

# 关键点 2: 使用 for 循环遍历返回的流式对象
for chunk in response_stream:
    # 检查每个数据块中是否有内容，并确保 choices 列表不为空
    if chunk.choices and chunk.choices[0].delta.content is not None:
        # 使用 end="" 和 flush=True 来实现打字机效果
        print(chunk.choices[0].delta.content, end="", flush=True)

# 在所有内容打印完毕后，输出一个换行符
print()

````
