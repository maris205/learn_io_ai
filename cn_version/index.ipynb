# 课程介绍

## ionet大模型服务简介

ionet 可以理解为一个GPU算力共享平台，对用户而言，用起来和vast.ai, autodl差不多，但其引入了去中心化GPU要素，相当于给算力供给方一些补贴，这样终端价格自然会便宜一些。

从用户的角度看，其有几个比较明显的优点；
* **开源大模型** ionet提供的大模型服务基本都是开源的，包括推理，训练等，这样非常适合学习、做原型使用。
* **免费服务** 大模型推理服务，RAG等基本给的token额度基本都是每天百万级的，基本够用了。
* **算力便宜** 即使需要总体价格和vast差不多，比aws等肯定是要便宜多了，

## ionet自己的介绍

当然，按照其自己的说法，其为人工智能和机器学习领域，特别是大模型服务，提供一个高性价比、可扩展且易于访问的计算平台。它并非直接提供自家的大语言模型，而是构建了一个底层基础设施，让开发者和企业能够经济高效地训练、微调和部署各类大型模型。

**核心理念：整合闲置GPU资源**

ionet的核心是其去中心化的物理基础设施网络（DePIN），它汇集了来自全球各地未被充分利用的GPU资源，包括独立数据中心、加密货币矿工以及个人贡献者。通过这种模式，ionet能够以远低于传统云服务提供商（如AWS、GCP）的价格提供强大的计算能力，成本节约可高达90%。

**为大模型服务提供的关键支持：**

* **模型推理与服务：** 部署训练好的大模型以提供实时服务（即模型推理）同样需要强大的计算支持。ionet的基础设施可以支持构建可扩展的模型推理工作流，确保应用程序能够快速、可靠地响应用户请求。开发者可以利用ionet的网络部署用于各种AI应用的模型，例如内容生成、图像识别等。

* **内置的RAG服务：** 为了方便开发者利用大模型构建知识库应用，ionet还提供了一项内置的检索增强生成（Retrieval-Augmented Generation, RAG）服务。用户可以通过网页界面或API轻松上传文档，构建自己的RAG知识库，从而让大模型能够基于特定领域的知识进行回答和交互。


* **模型训练与微调：** 大模型的训练需要巨大的并行计算能力。ionet的分布式GPU网络能够将复杂的训练任务分解并分配到成千上万个GPU上同时处理，显著缩短训练时间。这使得资源有限的初创公司和研究人员也能够进行大规模模型的训练和实验。


**主要优势：**

* **成本效益：** 通过利用闲置资源，ionet极大地降低了获取高性能计算能力的门槛。
* **可扩展性：** 其全球分布的庞大GPU网络可以根据用户需求即时扩展，从单个GPU到大规模集群，满足不同规模项目的需求。
* **易于访问：** ionet提供用户友好的网页界面和API，简化了GPU集群的部署和管理流程，并支持TensorFlow、PyTorch等主流机器学习框架。
* **灵活性：** 开发者可以根据自己的需求定制计算环境，并支持包括分布式训练、超参数调整和强化学习在内的多种复杂机器学习工作负载。

总而言之，ionet通过其去中心化的计算网络，为大模型服务的整个生命周期——从训练、微调到部署和推理——提供了一个强大而经济的解决方案，有力地推动了人工智能技术的普及和发展。

## 课程大纲
本课程主要是介绍ionet大模型服务的介绍和简单的例子，帮助大家能快速上手使用。

课程主要以ionet的服务为主线，分别介绍其io Intelligence和io cloud服务，

其中io Intelligence服务类似于saas服务，提供大模型的推理、RAG、Agent服务等

IO Cloud相当于算力租赁服务，可以通过ray、docker、虚拟机等形式租用算力。

我们对每个服务都给出一些入门的例子，简单实用为主。
