# Ionet agentic workflow

这里简单介绍一下“**Agentic Workflow**”（智能体工作流）。

我们可以把它想象成**普通员工**和**明星项目经理**之间的区别。

* **传统AI交互 (普通员工)**:
    你告诉他一个具体的指令，比如“帮我查一下今天的天气”，他完成并回复你。你必须一步一步地告诉他要做什么。

* **Agentic Workflow (明星项目经理)**:
    你只告诉他一个**宏大的目标**，比如“**帮我策划一次完美的周末东京旅行**”。然后，这位“项目经理”自己就会开始思考、规划、并执行一系列复杂的任务。

---

### 什么是Agentic Workflow？

**Agentic Workflow**（智能体工作流）是一种**由一个或多个自主的AI智能体（Agent）驱动的、用于完成复杂任务的高级工作模式**。

在这种模式下，AI不再只是被动地回答问题，而是变成了一个能够**主动思考、制定计划、使用工具、甚至相互协作**来达成最终目标的“行动者”。

简单来说，就是从“**你推一下，AI动一下**”变成了“**你设定目标，AI自己想办法搞定**”。

### Agentic Workflow vs. 传统提示词工程 (Prompt Engineering)

| 特性 | 传统提示词工程 (Prompt Engineering) | Agentic Workflow (智能体工作流) |
| :--- | :--- | :--- |
| **角色** | 是普通员工，完成具体任务 | **AI智能体**是项目经理，负责拆解任务 |
| **交互方式** | 一问一答，指令驱动 | 目标驱动，AI自主执行多步操作 |
| **复杂性** | 适合处理单一、明确的任务 | 适合处理复杂、开放式、需要多步推理的任务 |
| **工具使用** | 无法使用外部工具 | **可以自主调用工具**（如上网搜索、读写文件、执行代码） |



---

### 一个典型的Agentic Workflow示例

假设你的目标是：“**写一份关于‘自动驾驶汽车市场’的深度分析报告**”。

一个Agentic Workflow可能会这样运作：

1.  **规划与拆解 (Planning)**:
    * 主智能体（可以看作是“项目经理Agent”）接收到目标后，会把这个大任务拆解成几个子任务：
        1.  *任务A：* 上网搜索最新的市场规模数据、主要玩家和技术趋势。
        2.  *任务B：* 分析收集到的数据，找出关键的增长点和挑战。
        3.  *任务C：* 总结以上信息，撰写一份结构清晰的报告。

2.  **任务执行与工具调用 (Execution & Tool Use)**:
    * “项目经理Agent”可能会将任务分配给“专才Agent”：
        * 它命令一个“**研究员Agent**”去执行任务A。这个“研究员Agent”会自主调用**搜索引擎工具**，访问多个网站，抓取相关信息。
        * 然后，它命令一个“**数据分析师Agent**”去执行任务B。这个Agent可能会调用**代码执行工具**（如Python）来处理数据并生成图表。

3.  **反思与整合 (Reflection & Synthesis)**:
    * “项目经理Agent”收集所有“专才Agent”的工作成果。
    * 它会进行**自我反思**：信息是否全面？逻辑是否通顺？是否达成了最初的目标？如果发现不足，它可能会返回上一步，要求补充更多信息。
    * 最后，它将所有信息整合成一份连贯的、高质量的分析报告，呈现给你。

---

### Agentic Workflow的核心要素

* **自主性 (Autonomy)**：Agent能够独立做出决策，而无需人类每一步都进行干预。
* **规划能力 (Planning)**：能够将复杂目标分解为可执行的步骤。
* **工具使用 (Tool Use)**：能够调用外部API、数据库、代码解释器等来获取信息或执行操作。
* **记忆与反思 (Memory & Reflection)**：能够记住之前的操作和结果，并根据反馈进行调整和优化。
* **多智能体协作 (Multi-Agent Collaboration)**：多个拥有不同专长的Agent可以像一个团队一样协同工作。

总而言之，**Agentic Workflow代表了人机交互的未来范式**。它将我们从繁琐的具体指令中解放出来，让我们能够专注于“**做什么**”，而将“**怎么做**”的复杂过程，放心地交给一个由AI驱动的、高度自主的智能团队。

## ionet 工作流设计规则

基本规则如下：
1. 整体设计应该是按照agent mcp代码流程，设计的可视化编辑页面。
2. task是顶层节点，定义具体的输入需求。
3. 核心是Agent。Agent只能连接到task。
4. 工具tools是Agent的一部分。也就是tools只能连接到Agent。
5. 不同的task可以通过Stage连接，连接方式有串联，并联，循环等，类似与流程控制。
6. models也是Agent的一部分，默认用的应该是llama，如果设置了models，会使用设置的大模型替换默认的，当然，models也只能连接到Agent。

具体对应的yaml文件如下：

```
name: trans1
description: ''
client_mode: false
agents: []
tasks:
  - task_id: 9c7062a5-1323
    name: input
    objective: Translate "Hello, how are you?" to Spanish
    agents:
      - name: Translate Agent
        instructions: You are an assistant specialized in translation. {input}
        description: null
        tools: []
        memories: []
        model_settings:
          tool_choice: auto
    task_metadata: {}
    execution_metadata:
      execution_mode: sequential
      stages: []

```

## 工具使用

很多工作流平台，一般是把工具和agent当做一样的模块使用，有输入和输出，而io这里，工具只是Agent的一个部分，通过function call的形式调用。

一个获得当前时间的简单例子如下：

<img src='img/example3.png' width='600' />

这里同样需要注意的是，工具是Agent的一部分，需要Agent自己根据用户输入提示词，去判断什么时候去调用。因此，Agent大模型的提示词就写要通用一些，比如“你是一个小助手”之类的。而Task的任务提示词，也要写的明确一些，让大模型知道去调用tools。运行结果如下：

<img src='img/example4.png' width='600' />

如果把tool去掉，Agent会回答，他无法获得当前的时间。
具体对应的yaml文件如下：


```
name: get_time
description: ''
client_mode: false
agents: []
tasks:
  - task_id: b055fcbd-1098
    name: get_time
    objective: 现在几点了
    agents:
      - name: Solar
        instructions: You are a helpful assistant.
        description: null
        tools:
          - what_time_is_it
        memories: []
        model_settings:
          tool_choice: auto
    task_metadata: {}
    execution_metadata:
      execution_mode: sequential
      stages: []

```
