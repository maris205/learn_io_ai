## Ray框架简介

### 一句话概括

**Ray 是一个开源的、通用的分布式计算框架**，其核心目标是让开发者能够**极其简单地将单机 Python 程序扩展到大规模的计算集群上**运行，而无需关心复杂的底层系统问题。

你可以把它理解为\*\*“分布式 Python 的操作系统”\*\*。

-----

### Ray 解决了什么痛点？

1.  **Python 的并行限制**：Python 因为全局解释器锁（GIL）的存在，在单台机器上很难真正利用好多核 CPU 进行计算密集型并行处理。
2.  **分布式编程的复杂性**：当一台机器的计算资源不够用，需要扩展到多台机器（集群）时，开发者就需要面对网络通信、任务调度、节点失败、状态管理等一系列极其复杂的问题。这通常需要专门的分布式系统工程师来处理。

**Ray 的出现，就是为了让你像写普通 Python 代码一样，轻松地驾驭整个计算集群的强大算力。**

-----

### Ray 的两个核心概念

Ray 的魔力主要来自于它提供的两个简单而强大的基础模块（API）：

1.  **Tasks (远程任务)**：

      * **做什么用**：将一个普通的 Python **函数** 变成可以在集群中任何一个节点上**异步执行**的任务。
      * **怎么实现**：只需要在你的函数上加一个装饰器 `@ray.remote`。
      * **如何调用**：调用时，将 `my_function()` 改为 `my_function.remote()`。
      * **特点**：这个调用会立即返回一个“未来对象”（Object ID），程序会继续往下走，而 Ray 会在后台的某个节点上调度并执行这个函数。你可以稍后通过 `ray.get()` 获取真正的结果。这使得大规模的并行计算变得异常简单。

2.  **Actors (远程演员)**：

      * **做什么用**：将一个普通的 Python **类** 变成一个可以在集群中存活的、有状态的服务。
      * **怎么实现**：同样，在你的类上加一个装饰器 `@ray.remote`。
      * **如何调用**：创建实例时，将 `MyClass()` 改为 `MyClass.remote()`。
      * **特点**：Actor 非常适合那些需要维护自身状态的场景，比如一个环境模拟器、一个参数服务器、或者一个加载了很大模型并需要反复被调用的推理服务。

-----

### Ray 的主要用途和生态系统

Ray 不仅仅是一个核心框架，它还拥有一个强大的原生库生态系统，专门用于解决机器学习和AI领域的各种大规模问题：

  * **Ray Tune**：分布式**超参数调优**。可以同时启动成百上千个不同参数的训练任务，并进行高效管理。
  * **Ray Train**：分布式**模型训练**。可以轻松地将 PyTorch、TensorFlow 等单机训练代码扩展到多台机器、多个 GPU 上进行并行训练。
  * **Ray Serve**：分布式**模型服务**。能够将训练好的模型部署成一个高可用、可扩展的在线推理服务（API）。
  * **Ray Data**：分布式**数据处理**。为大规模机器学习任务提供数据加载和预处理的能力。

这些库共同构成了一个端到端的、覆盖 AI 应用全生命周期的分布式解决方案。


### 小结

Ray 极大地降低了分布式编程的门槛，让广大的 Python 开发者和数据科学家能够专注于算法和应用逻辑本身，而不是底层的分布式系统细节。如果你需要处理的任务超出了单台机器的处理能力，无论是数据处理、模型训练还是超参数搜索，Ray 都是一个非常值得学习和使用的强大工具。

Ray 框架作为一个通用的分布式计算平台，已成为大模型生态的核心基础设施，它与 Hugging Face Transformers 和 DeepSpeed 等关键工具实现了无缝集成。开发者可以利用 Ray，将为单 GPU 编写的标准 Hugging Face 训练代码几乎无需修改，便能轻松扩展至大规模集群上运行。同时，Ray 还能作为“指挥官”来调度 DeepSpeed，通过其强大的内存优化技术，实现对远超单卡显存容量的巨型模型的分布式训练。总而言之，Ray 屏蔽了底层的分布式复杂性，让开发者能在一个统一的平台上，便捷地组合使用这些业界最顶尖的工具，从而极大地简化了大规模AI模型的开发和部署。

### hello world 代码

注意，下面代码请在部署的ray环境中运行

```python
#对照组，普通的 Python 代码 (串行执行，耗时约4秒):

import time

def slow_function(i):
    time.sleep(1)
    return i

results = []
for i in range(4):
    results.append(slow_function(i))

print(results)
```

```python
#使用 Ray 的代码 (并行执行，耗时约1秒):
import ray
import time

# 启动 Ray
ray.init()

# 加上 @ray.remote 装饰器
@ray.remote
def slow_function(i):
    time.sleep(1)
    return i

# 使用 .remote() 异步调用
futures = [slow_function.remote(i) for i in range(4)]

# 使用 ray.get() 获取所有结果
results = ray.get(futures)

print(results)
```

可以看到，ray编程非常简单，只需要:

1. 把之前的函数加上@ray.remote装饰器即可
2. 函数.remote() 调用远程worker工作
3. ray.get() 阻塞式等待结果

### gpu 测试例子 

```python
#对照组，本地服务器执行
import ray
import torch
import os

def check_gpu_on_ray_worker():
    """
    这个函数会在一个独立的 Ray 工作进程 (Worker) 中执行。
    Ray 调度器会确保这个 Worker 能够访问到至少一个 GPU。
    """
    print("--- Log from the Ray Worker process ---")
    
    # 检查 CUDA 是否真的可用
    if not torch.cuda.is_available():
        error_message = "Error: Ray assigned a GPU, but PyTorch cannot detect it."
        print(error_message)
        return {"error": error_message}

    # 获取 GPU 信息
    gpu_index = torch.cuda.current_device()
    gpu_name = torch.cuda.get_device_name(gpu_index)

    # 将检查结果打包成一个字典返回
    result = {
        "pytorch_version": torch.__version__,
        "cuda_available": True,
        "gpu_index": gpu_index,
        "gpu_name": gpu_name,
        "worker_pid": os.getpid()
    }
    return result


gpu_info = check_gpu_on_ray_worker()

print(gpu_info)
```

可以看到，打印结果如下，这个是当前登录服务器的信息，并不是我们的4090工作服务器

```python
import ray
import torch
import os

# 步骤 1: 将 GPU 检查代码封装成一个函数
# 步骤 2: 使用 @ray.remote 装饰器将其变为 Ray Task
# 关键: 使用 num_gpus=1 来告诉 Ray，这个任务必须在有 GPU 的节点上运行
@ray.remote(num_gpus=1)
def check_gpu_on_ray_worker():
    """
    这个函数会在一个独立的 Ray 工作进程 (Worker) 中执行。
    Ray 调度器会确保这个 Worker 能够访问到至少一个 GPU。
    """
    print("--- Log from the Ray Worker process ---")
    
    # 检查 CUDA 是否真的可用
    if not torch.cuda.is_available():
        error_message = "Error: Ray assigned a GPU, but PyTorch cannot detect it."
        print(error_message)
        return {"error": error_message}

    # 获取 GPU 信息
    gpu_index = torch.cuda.current_device()
    gpu_name = torch.cuda.get_device_name(gpu_index)
    
    # 打印 Worker 端的日志，方便调试
    print(f"PyTorch version: {torch.__version__}")
    print(f"CUDA available: True")
    print(f"Worker process PID: {os.getpid()}")
    print(f"Running on GPU Index: {gpu_index}")
    print(f"GPU Name: {gpu_name}")
    print("------------------------------------")

    # 将检查结果打包成一个字典返回
    result = {
        "pytorch_version": torch.__version__,
        "cuda_available": True,
        "gpu_index": gpu_index,
        "gpu_name": gpu_name,
        "worker_pid": os.getpid()
    }
    return result


# 初始化 Ray，不需要多次
ray.init()

# 使用 .remote() 异步执行任务
# Ray 会找到一个满足 num_gpus=1 要求的 Worker 来运行它
check_future = check_gpu_on_ray_worker.remote()

# 步骤 5: 使用 ray.get() 获取任务的返回结果
# 这会阻塞程序，直到远程函数执行完毕
gpu_info = ray.get(check_future)

# 以更美观的方式打印从 Worker 返回的字典
print(result)
```

### 大模型微调例子

这里使用ray官方给出的基于bert大模型进行分类微调的例子，分别给出本地运行和ray运行的代码

```python
#本地运行的例子，可以基于ionet vm虚拟机运行，不要在ray的登录服务器上运行！！！会把任务都卡主的
#使用ionet的vm，需要安装下面的包，在jupyter里面安装即可
# pip install transformers[torch] 
# pip install tf_keras evaluate


# Adapted from Hugging Face tutorial: https://huggingface.co/docs/transformers/training

import numpy as np
import evaluate
from datasets import load_dataset
from transformers import (
    Trainer,
    TrainingArguments,
    AutoTokenizer,
    AutoModelForSequenceClassification,
)

# Datasets
dataset = load_dataset("yelp_review_full")
tokenizer = AutoTokenizer.from_pretrained("bert-base-cased")

def tokenize_function(examples):
    return tokenizer(examples["text"], padding="max_length", truncation=True)

small_train_dataset = dataset["train"].select(range(10000)).map(tokenize_function, batched=True)
small_eval_dataset = dataset["test"].select(range(10000)).map(tokenize_function, batched=True)

# Model
model = AutoModelForSequenceClassification.from_pretrained(
    "bert-base-cased", num_labels=5
)

# Metrics
metric = evaluate.load("accuracy")

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)
    return metric.compute(predictions=predictions, references=labels)

# Hugging Face Trainer
training_args = TrainingArguments(
    output_dir="test_trainer"
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=small_train_dataset,
    eval_dataset=small_eval_dataset,
    compute_metrics=compute_metrics,
)

# Start Training
trainer.train()
```

```python
import os
import pyarrow.fs  # Ensure pyarrow is installed (pip install pyarrow)
import numpy as np
import evaluate
from datasets import load_dataset
from transformers import (
    Trainer,
    TrainingArguments,
    AutoTokenizer,
    AutoModelForSequenceClassification,
)

import ray.train.huggingface.transformers
from ray.train import ScalingConfig, RunConfig
from ray.train.torch import TorchTrainer

import ray


"""
如果是4090等游戏多卡，需要下面的设置
如果是h100，v100等专业卡，则不需要设置
"""
# Initialize Ray with runtime_env to upgrade pyarrow on workers
ray.init(
    runtime_env={
        "env_vars": {  # Add environment variables here for workers
            "NCCL_P2P_DISABLE": "1",
            "NCCL_IB_DISABLE": "1"
        }
    }
)

# [1] Encapsulate data preprocessing, training, and evaluation
# logic in a training function
# ============================================================
def train_func():
    # Datasets
    dataset = load_dataset("yelp_review_full")
    tokenizer = AutoTokenizer.from_pretrained("bert-base-cased")

    def tokenize_function(examples):
        return tokenizer(examples["text"], padding="max_length", truncation=True)

    small_train_dataset = (
        dataset["train"].select(range(10000)).map(tokenize_function, batched=True)
    )
    small_eval_dataset = (
        dataset["test"].select(range(10000)).map(tokenize_function, batched=True)
    )

    # Model
    model = AutoModelForSequenceClassification.from_pretrained(
        "bert-base-cased", num_labels=5
    )

    # Evaluation Metrics
    metric = evaluate.load("accuracy")

    def compute_metrics(eval_pred):
        logits, labels = eval_pred
        predictions = np.argmax(logits, axis=-1)
        return metric.compute(predictions=predictions, references=labels)

    # Hugging Face Trainer
    training_args = TrainingArguments(
        output_dir="test_trainer",
        evaluation_strategy="epoch",
        save_strategy="epoch",
        report_to="none",
    )

    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=small_train_dataset,
        eval_dataset=small_eval_dataset,
        compute_metrics=compute_metrics,
    )

    # [2] Report Metrics and Checkpoints to Ray Train
    # ===============================================
    callback = ray.train.huggingface.transformers.RayTrainReportCallback()
    trainer.add_callback(callback)

    # [3] Prepare Transformers Trainer
    # ================================
    trainer = ray.train.huggingface.transformers.prepare_trainer(trainer)

    # Start Training
    trainer.train()


# Create custom S3 filesystem with credentials
fs = pyarrow.fs.S3FileSystem(
    access_key="******",  # Replace with your actual key
    secret_key="******",  # Replace with your actual secret
    region="us-west-1",  # Specify your bucket's region here
    # Optional: Add 'session_token="YOUR_SESSION_TOKEN"' if using temporary creds
    # Optional: Add 'endpoint_override="https://s3.custom-endpoint.com"' for non-AWS S3-compatible storage like MinIO
)

# [4] Define a Ray TorchTrainer to launch `train_func` on all workers
# ===================================================================
ray_trainer = TorchTrainer(
    train_func,
    scaling_config=ScalingConfig(num_workers=2, use_gpu=True),
    run_config=RunConfig(
        # Use bucket/path without the 's3://' prefix when providing a custom filesystem
        storage_path="ionet2/ray1",
        storage_filesystem=fs,
        # Optional: Add 'name="your-experiment-name"' to organize runs
    ),
)
result: ray.train.Result = ray_trainer.fit()

# [5] Load the trained model
with result.checkpoint.as_directory() as checkpoint_dir:
    checkpoint_path = os.path.join(
        checkpoint_dir,
        ray.train.huggingface.transformers.RayTrainReportCallback.CHECKPOINT_NAME,
    )
    model = AutoModelForSequenceClassification.from_pretrained(checkpoint_path)
```
