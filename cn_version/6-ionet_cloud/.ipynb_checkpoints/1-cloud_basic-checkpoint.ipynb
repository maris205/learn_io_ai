# ionet Cloud服务简介

## io.net Cloud 核心服务

`io.net Cloud` 提供了一系列灵活的部署选项，旨在满足从个人开发者到企业级用户的不同AI计算需求。其主要服务包括：

1.  **容器引擎 (Container Engine)**
    * **简介**: 这是一个为现代开发者设计的服务，允许用户将自己的Docker容器直接部署到 `io.net` 的高性能GPU集群上。它极大地简化了环境配置，是快速启动模型训练和AI应用最便捷的方式。

2.  **虚拟机 (Virtual Machine)**
    * **简介**: 提供一个预装了主流AI框架（如PyTorch, TensorFlow）的完整虚拟机环境，就像使用AWS EC2一样。它非常适合需要持久化、可交互开发环境的用户，例如运行Jupyter Notebook进行数据探索和模型调试。

3.  **Ray集群 (Ray Cluster)**
    * **简介**: 为大规模分布式Python和AI应用开发者量身定制。用户可以一键部署一个去中心化的Ray集群，轻松地将复杂的分布式训练或超参数调优任务，扩展到 `io.net` 全球的GPU网络上。

4.  **按需裸金属 (Bare Metal on demand)**
    * **简介**: 面向需要极致性能和硬件控制权的专家用户。该服务提供对物理服务器（裸金属）的直接访问，消除了虚拟化层的性能开销，适用于HPC（高性能计算）等要求严苛的任务。

5.  **Kubernetes/暂未上线 (Kubernetes Cluster)**
    * **简介**: 一项面向企业的服务，允许将 `io.net` 的去中心化GPU资源作为计算节点，无缝集成到现有的Kubernetes工作流中。这使得企业可以利用 `io.net` 的低成本算力，弹性扩展其云原生AI应用。

**总结来说**，`io.net Cloud` 通过拥抱Docker、虚拟机、Ray和Kubernetes等行业标准，为用户提供了一个**低成本、大规模、且高度灵活**的去中心化AI计算平台。


<img src='img/cloud1.png' width='600' />

对用户而言，大致可以分为两类服务：
1. 专业服务。包括Ray集群和按需裸金属，这两个以训练大规模的模型为主，需要比较专业的大模型知识，如ray的学习等，一般是开多个机器做集群，然后跑训练服务。
2. 学习开发。包括容器引擎和虚拟机，主要是单机/单GPU为主，和autodl，vast.ai等提供的服务类似，适合学习和开发使用，做基本的微调也是可以的。

## 服务费用

ionet的cloud服务，相当于是GPU租赁中介服务，是需要给GPU提供方支付费用的，因此，这一部分是收费的，但费用相对比较便宜，特别是和AWS等比。

另外，因为相当于是C2C的租赁服务，也可以租用4090等游戏卡，也是不违反NVIDIA的规定的。

支付方面，目前主要接受io的token或者usdt的token支付，需要的可以到交易所购买

也支持一般的visa，master等信用卡支付

租用的价格整体比较低，但会根据供给和需求有小幅的调整，机器数量和价格可以从页面：

https://explorer.io.net/explorer/home

查看。

对一个初创公司，能提供几千个卡的量级，而且价格便宜，已经非常不错了。
