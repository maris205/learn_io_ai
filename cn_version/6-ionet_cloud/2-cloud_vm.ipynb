# ionet cloud 虚拟机

ionet的虚拟机和一般的aws的ecs服务器等都是类似的，是最常用的一种服务器租赁方式

针对大模型的应用，都预装了预装 PyTorch 和 CUDA 等框架，可以直接运行大部分的机器学习代码。

具体租用流程和aws等云平台类似，这里主要说一些要点。

虚拟机的服务为：
https://cloud.io.net/cloud/virtual-machines

<img src='img/cloud2.png' width='600' />

点击deploy。。 那个按钮即可开始部署

## 步骤1 网速确认

选择默认即可

<img src='img/netspeed.png' width='600' />

## 步骤2 选择服务器类型

学习、测试一般选择4090即可

注意需要点到显卡类型上，然后选择服务器的地理位置：

<img src='img/card.png' width='600' />

## 步骤3 确定镜像添加ssh key

这一步主要是确定使用哪个镜像，一般选择第二个Data science...即可，自带jupyter 服务

<img src='img/ssh.png' width='600' />

软件环境如下：
- OS: Ubuntu 24.04.1 LTS (Noble Numbat)
- CUDA: 12.1
- RAPIDS: 25.6.0
- Python: 3.12
- Conda: 25.7.0

然后ssh key，简单的，就是添加自己的pc上的pub公钥内容即可：

<img src='img/key.png' width='600' />


## 步骤4 添加额外网络端口

这用过aws等服务器的都知道，就是开额外的网络端口，默认的一般只有22等常用端口

像我们使用jupyter，端口就是8888，需要专门添加下，要不然无法访问

另外需要注意的是，服务器创建后，就没法再添加了，只能是创建的时候就添加

<img src='img/port.png' width='600' />

## 步骤5 部署确认

这一步就是确认，主要是使用时间，测试一般选1个小时即可

另外需要选择支付方式，有多种支付方式可选，包括传统的和区块链的

<img src='img/sub.png' width='600' />

确认后到支付页面支付确认就开始部署了，一般需要几分钟的样子，就能部署成功了：

<img src='img/pay.png' width='600' />

成功后：
<img src='img/view.png' width='600' />

## 启动jupyter

首先启动jupyter，运行jupyter启动脚本即可：
```
launch-jupyter
```


<img src='img/jupyter.png' width='600' />



启动时候的，可以看到启动日志中的登录token，这个就是类似这样的

http://127.0.0.1:8888/lab?token=1981227a4100db27f884280cb7494fdf8952481d0313a379

token后面的参数 1981227a4100db27f884280cb7494fdf8952481d0313a379 就是登录token，这个和一般的jupyter使用是类似的。


## 使用jupyter

首先要确定juypter的外网地址，这个地址和ssh登录的地址是一样的，如ssh地址为
52.205.25.210

juypter登录地址也是52.205.25.210

然后是端口，jupyter启动的服务端口为8888，我们在网络服务也添加了这个端口，但需要注意的是，这个是内网端口

安全期间，ionet会映射一个外网端口，具体点network service，选择juypter服务就可以看到：

<img src='img/network.png' width='600' />

这里端口8888对应的外网端口就是31814

然后jupyter的访问地址就是：
http://52.205.25.210:31814

<img src='img/jplog.png' width='600' />

需要填写的token就是上一步获得的1981227a4100db27f884280cb7494fdf8952481d0313a379

填写后点login即可：

<img src='img/jp.png' width='600' />


另外需要注意的是，jupyter的web登录页面是有websocket服务的，因此，访问的时候最好不要用vpn和代理，否则会非常非常慢，甚至访问不了

注意下即可。登录后可以用vpn问题不大。

## 环境测试

可以使用如下代码测下环境是否正常：

```
import torch

print(f"PyTorch version: {torch.__version__}")

# 检查 CUDA 是否被 PyTorch 检测到
is_available = torch.cuda.is_available()

print(f"Is CUDA available? {is_available}")

# 获取当前默认 GPU 的索引
current_device_index = torch.cuda.current_device()
print(f"Current GPU index: {current_device_index}")


# 获取当前 GPU 的名称
current_device_name = torch.cuda.get_device_name(current_device_index)
print(f"Current GPU name: {current_device_name}")

```
