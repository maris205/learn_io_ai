# Ionet Training Service

Ionet provides model training services in the form of SAAS, which is Training as a Service (TaaS). You can use your own data to fine-tune existing models to achieve customized performance without training from scratch.

This tool is designed for builders, researchers, developers, and enterprises who want better control over AI behavior and evolution. With support for advanced training techniques such as PPO, DPO, and reward models, you can go beyond traditional tuning methods and explore cutting-edge machine learning technologies.

<img src='img/train1.png' width='600' />

Currently still in the testing phase, quota application is required to use it. Here we mainly explain the basic usage.

## Step 1: Choose Training Method

This refers to the form of large model training. Different stages and purposes require different methods. The main types are:

- Supervised Fine-tuning: Use labeled datasets to train your model for specific task learning.
- Reward Modeling: Train your model to assign scores to generated responses.
- Proximal Policy Optimization (PPO): Use reinforcement learning with reward feedback.
- Direct Preference Optimization (DPO): Directly optimize the model using ranking preferences.
- Controlled Tuning Optimization (Experimental): Apply KL regularization tuning for fine-grained control.

<img src='img/train2.png' width='600' />

The most commonly used is the first one, supervised learning, including classic machine learning method training and instruction fine-tuning. This is sufficient when building large model applications.
The subsequent methods are generally used when training large models.



## Step 2: Choose Base Model

This means selecting a pre-trained model. Currently, you can directly choose from some default models or use Huggingface model link addresses.

<img src='img/train3.png' width='600' />

Try to choose from the default list, as they should be better supported.

## Step 3: Choose Dataset

This means selecting training data. Similar to the previous step, you can choose from some default datasets or directly use Huggingface dataset addresses.

<img src='img/train4.png' width='600' />

Currently, you can only choose from the default list. Huggingface data links are not yet available.
