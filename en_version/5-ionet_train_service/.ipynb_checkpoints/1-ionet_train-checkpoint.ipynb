# ionet训练服务

ionet提供了SAAS形式的模型训练服务，也就是训练即服务 (TaaS)，您可以使用自己的数据微调现有模型 - 实现定制性能，而无需从头开始训练。

此工具专为希望更好地控制 AI 行为和演进的构建者、研究人员、开发者和企业而设计。借助 PPO、DPO 和奖励模型等高级训练技术的支持，您可以超越传统的调优方式，尝试机器学习的前沿技术。

<img src='img/train1.png' width='600' />

目前还处在测试阶段，需要申请额度才能使用，这里主要说下其基本的用法。

## 第一步：选择训练方法

就是大模型训练的形式，不同阶段，不同目的选择不同的即可，主要有以下几种：

- 监督微调：使用标记数据集来训练您的模型，以进行特定任务的学习。
- 奖励建模：训练您的模型为生成的响应分配分数。
- 近端策略优化（PPO）： 使用带有奖励反馈的强化学习。
- 直接偏好优化 (DPO)：使用排序偏好直接优化模型。
- 受控调整优化 （实验）：应用 KL 正则化调整进行细粒度控制。

<img src='img/train2.png' width='600' />


最常用的就是第一个监督学习，包括经典的机器学习方法训练，以及指令微调等，做大模型应用的时候都用这个就行。
后续几个一般是训练大模型的时候使用的。



## 第二步：选择基础模型

就是选择预训练模型，目前可以直接选择默认的一些模型，也可以直接使用Huggingface的模型链接地址

<img src='img/train3.png' width='600' />

尽量选择默认列表的，应该是支持的更好一些。

## 第三步：选择数据集

就是选择训练数据，这个和前面一样的，可以选默认的一些数据集，也可以直接使用Huggingface的数据集地址

<img src='img/train4.png' width='600' />

目前只能选择默认列表的，Huggingface的数据链接还用不了


选择完成后，就进入训练环节，目前是需要申请一些额度才能使用，不是完全开放的

后续等完全开放后再做后续的讲解
