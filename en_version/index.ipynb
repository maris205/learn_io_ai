# Course Introduction

## Introduction to ionet Large Model Service

ionet can be understood as a GPU computing power sharing platform. For users, it works similarly to vast.ai and autodl, but it introduces the concept of decentralized GPU elements, which provides subsidies to computing power suppliers, thus making the end price naturally lower.

From the user's perspective, it has several obvious advantages;
* **Open Source Large Models**: The large model services provided by ionet are basically open source, including inference, training, etc., which is very suitable for learning and prototyping.
* **Free Services**: The token quota provided for large model inference services, RAG, etc. is basically on the order of millions per day, which is basically sufficient.
* **Affordable Computing Power**: Even if the total price is similar to vast, it is definitely cheaper than AWS, etc.

## ionet's Own Introduction

Of course, according to its own statement, it provides a cost-effective, scalable and easily accessible computing platform for the field of artificial intelligence and machine learning, especially large model services. It does not directly provide its own large language models, but builds an underlying infrastructure that allows developers and enterprises to economically and efficiently train, fine-tune and deploy various large models.

**Core Concept: Integrating Idle GPU Resources**

The core of ionet is its decentralized physical infrastructure network (DePIN), which brings together underutilized GPU resources from around the world, including independent data centers, cryptocurrency miners and individual contributors. Through this model, ionet is able to provide powerful computing capabilities at a price much lower than traditional cloud service providers (such as AWS, GCP), with cost savings of up to 90%.

**Key Support for Large Model Services:**

* **Model Inference and Services:** Deploying trained large models to provide real-time services (i.e., model inference) also requires powerful computing support. ionet's infrastructure can support the construction of scalable model inference workflows, ensuring that applications can quickly and reliably respond to user requests. Developers can use ionet's network to deploy models for various AI applications, such as content generation, image recognition, etc.

* **Built-in RAG Service:** To facilitate developers in using large models to build knowledge base applications, ionet also provides a built-in Retrieval-Augmented Generation (RAG) service. Users can easily upload documents through a web interface or API to build their own RAG knowledge base, allowing large models to answer and interact based on domain-specific knowledge.


* **Model Training and Fine-tuning:** Training large models requires enormous parallel computing power. ionet's distributed GPU network can decompose complex training tasks and distribute them to tens of thousands of GPUs for simultaneous processing, significantly reducing training time. This enables startups and researchers with limited resources to also conduct large-scale model training and experiments.


**Main Advantages:**

* **Cost-effectiveness:** By utilizing idle resources, ionet greatly reduces the threshold for accessing high-performance computing capabilities.
* **Scalability:** Its globally distributed massive GPU network can be instantly scaled according to user needs, from a single GPU to a large-scale cluster, meeting the needs of projects of different scales.
* **Easy Access:** ionet provides user-friendly web interfaces and APIs, simplifying the deployment and management of GPU clusters, and supporting mainstream machine learning frameworks such as TensorFlow and PyTorch.
* **Flexibility:** Developers can customize computing environments according to their own needs and support various complex machine learning workloads including distributed training, hyperparameter tuning and reinforcement learning.

In summary, through its decentralized computing network, ionet provides a powerful and economical solution for the entire life cycle of large model services - from training, fine-tuning to deployment and inference - strongly promoting the popularization and development of artificial intelligence technology.

## Course Outline
This course mainly introduces the ionet large model service and provides simple examples to help everyone quickly get started with using it.

The course mainly focuses on ionet's services, introducing its io Intelligence and io Cloud services respectively.

Among them, io Intelligence service is similar to SaaS service, providing large model inference, RAG, Agent services, etc.

IO Cloud is equivalent to computing power rental service, which can rent computing power through ray, docker, virtual machine, etc.

We provide some introductory examples for each service, focusing on simplicity and practicality.
