# Ionet Agentic Workflow

Here's a brief introduction to "**Agentic Workflow**" (Intelligent Agent Workflow).

We can think of it as the difference between a **regular employee** and a **star project manager**.

* **Traditional AI Interaction (Regular Employee)**:
    You give them a specific instruction, like "Help me check today's weather", they complete it and reply to you. You must tell them step by step what to do.

* **Agentic Workflow (Star Project Manager)**:
    You only tell them a **grand goal**, like "**Help me plan a perfect weekend trip to Tokyo**". Then, this "project manager" will start thinking, planning, and executing a series of complex tasks on their own.

---

### What is Agentic Workflow?

**Agentic Workflow** (Intelligent Agent Workflow) is a **high-level working mode driven by one or more autonomous AI intelligent agents (Agents) to complete complex tasks**.

In this mode, AI is no longer just passively answering questions, but becomes an "actor" capable of **active thinking, planning, using tools, and even collaborating with each other** to achieve the final goal.

Simply put, it changes from "**you push once, AI moves once**" to "**you set the goal, AI figures out how to get it done**".

### Agentic Workflow vs. Traditional Prompt Engineering

| Feature | Traditional Prompt Engineering | Agentic Workflow (Intelligent Agent Workflow) |
| :--- | :--- | :--- |
| **Role** | Regular employee, completing specific tasks | **AI Agent** is the project manager, responsible for task breakdown |
| **Interaction** | Q&A, instruction-driven | Goal-driven, AI autonomously executes multi-step operations |
| **Complexity** | Suitable for single, clear tasks | Suitable for complex, open-ended tasks requiring multi-step reasoning |
| **Tool Usage** | Cannot use external tools | **Can autonomously call tools** (such as web search, file read/write, code execution) |

---

### A Typical Agentic Workflow Example

Suppose your goal is: "**Write a deep analysis report on the 'autonomous vehicle market'**".

An Agentic Workflow might operate like this:

1. **Planning & Breakdown**:
   * The main agent (can be seen as a "Project Manager Agent") receives the goal and breaks down this large task into several subtasks:
      1. *Task A:* Search the web for the latest market size data, major players, and technology trends.
      2. *Task B:* Analyze the collected data to identify key growth points and challenges.
      3. *Task C:* Summarize the above information and write a well-structured report.

2. **Task Execution & Tool Usage**:
   * The "Project Manager Agent" might assign tasks to "specialist agents":
       * It commands a "**Researcher Agent**" to execute Task A. This "Researcher Agent" will autonomously call **search engine tools**, visit multiple websites, and scrape relevant information.
       * Then, it commands a "**Data Analyst Agent**" to execute Task B. This Agent might call **code execution tools** (like Python) to process data and generate charts.

3. **Reflection & Synthesis**:
   * The "Project Manager Agent" collects all work results from "specialist agents".
   * It performs **self-reflection**: Is the information comprehensive? Is the logic coherent? Has the original goal been achieved? If deficiencies are found, it might return to the previous step and request additional information.
   * Finally, it integrates all information into a coherent, high-quality analysis report and presents it to you.

---

### Core Elements of Agentic Workflow

* **Autonomy**: Agents can make independent decisions without human intervention at every step.
* **Planning Capability**: Can break down complex goals into executable steps.
* **Tool Usage**: Can call external APIs, databases, code interpreters, etc., to obtain information or execute operations.
* **Memory & Reflection**: Can remember previous operations and results, and adjust and optimize based on feedback.
* **Multi-Agent Collaboration**: Multiple agents with different expertise can work together like a team.

In summary, **Agentic Workflow represents the future paradigm of human-computer interaction**. It liberates us from tedious specific instructions, allowing us to focus on "**what to do**", while entrusting the complex process of "**how to do it**" to an AI-driven, highly autonomous intelligent team.


## Ionet Agentic Workflow

Ionet provides a web-based visual workflow design interface and detailed SDK. This section provides some examples of web-based design.

Web interface address: https://ai.io.net/ai/agentic-workflow-editor

<img src='img/work_flow.png' width='600' />

## Ionet Workflow Design Rules

The basic rules are as follows:
1. The overall design should be a visual editing interface designed according to the agent MCP code flow.
2. Task is the top-level node that defines specific input requirements.
3. The core is Agent. Agents can only connect to tasks.
4. Tools are part of the Agent. That is, tools can only connect to Agents.
5. Different tasks can be connected through Stages, with connection methods including series, parallel, loops, etc., similar to flow control.
6. Models are also part of the Agent. The default should be llama. If models are set, the set large model will be used to replace the default. Of course, models can only connect to Agents.

## Translation Example

Based on these rules, the translation example is as follows:

<img src='img/example1.png' width='600' />

It should be noted that the task contains a complete task prompt written explicitly, rather than just the content to be translated like in general workflows. After running, the result output can be obtained:

<img src='img/example2.png' width='600' />

The corresponding YAML file is as follows:

```
name: trans1
description: ''
client_mode: false
agents: []
tasks:
  - task_id: 9c7062a5-1323
    name: input
    objective: Translate "Hello, how are you?" to Spanish
    agents:
      - name: Translate Agent
        instructions: You are an assistant specialized in translation. {input}
        description: null
        tools: []
        memories: []
        model_settings:
          tool_choice: auto
    task_metadata: {}
    execution_metadata:
      execution_mode: sequential
      stages: []

```

## Tool Usage

Many workflow platforms generally treat tools and agents as the same type of module with inputs and outputs. However, in io, tools are merely a component of an Agent and are invoked via function calls.

A simple example for retrieving the current time is shown below:

<img src='img/example3.png' width='600' />

It's important to note here that tools are part of the Agent. The Agent must determine, based on the user's input prompt, when to invoke a tool. Therefore, the Agent's prompt should be written in a general way, such as "You are a helpful assistant." Meanwhile, the task prompt should be specific enough to guide the large model to call the appropriate tools. The execution result is shown below:

<img src='img/example4.png' width='600' />

If the tool is removed, the Agent would respond that it cannot obtain the current time. The corresponding YAML file is as follows:

```
name: get_time
description: ''
client_mode: false
agents: []
tasks:
  - task_id: b055fcbd-1098
    name: get_time
    objective: What time is it now?
    agents:
      - name: Solar
        instructions: You are a helpful assistant.
        description: null
        tools:
          - what_time_is_it
        memories: []
        model_settings:
          tool_choice: auto
    task_metadata: {}
    execution_metadata:
      execution_mode: sequential
      stages: []
```

Other tool invocations follow a similar pattern. However, search-related tools and flow control tools have not been successfully debugged yet. These should be tested again once the official io documentation becomes available.

Currently, io's workflow is still under continuous development. Overall, based on the current design approach, it's more suitable for Agent developers. For beginners who don't understand code, it may not be as easy to get started with as common workflow orchestration patterns.
