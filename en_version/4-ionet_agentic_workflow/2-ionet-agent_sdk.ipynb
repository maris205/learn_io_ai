The iointel library provides a flexible system for building and orchestrating agents and workflows. It offers two modes:

- Client mode: Tasks call remote API clients (e.g., your client.py functionality).
- Local mode: Tasks run directly in the local environment, utilizing run_agents(...) local logic.
It also supports loading YAML or JSON workflows to define multi-step tasks.

Local mode execution

```python
tasks = Workflow(objective="Breaking news: local sports team wins!", client_mode=False)
await tasks.summarize_text(max_words=50).run_tasks()
```

Remote execution

```python
tasks = Workflow(objective="Breaking news: local sports team wins!", client_mode=True)
await tasks.summarize_text(max_words=50).run_tasks()
```

Workflow chain, where multiple agents work in series

```python
tasks = Workflow("Tech giant acquires startup for $2B", client_mode=False)
(tasks
   .translate_text(target_language="spanish")
   .sentiment()
)
await  tasks.run_tasks()
```

### Agentic Tools with Beautiful User Interface

Features
Agentic tool usage: Agents can call Python tools, return results, and perform chained reasoning.

- Rich tool call visualization: Tool calls and results are presented as beautiful golden "pills" in both CLI (rich) and Gradio UI.
- Dynamic UI: Agents can dynamically generate forms (text boxes, sliders, etc.) in Gradio applications.
- Real-time CSS themes: Agents can change UI themes at runtime.
- Jupyter compatible: Gradio UI can be launched in notebook cells.


```python
from iointel import Agent, register_tool

@register_tool
def add(a: float, b: float) -> float:
    return a + b

agent = Agent(
    name="Solar",
    instructions="You are a helpful assistant.",
    model="openai/gpt-oss-120b",
    tools=[add],
    show_tool_calls=True,  # Pretty rich tool call output!
    api_key=api_key,
    base_url="https://api.intelligence.io.solutions/api/v1"
)

import asyncio
async def main():
    result = await agent.run("What is 2 + 2?", pretty=True)
    # Tool calls/results are shown in rich formatting!

await main()
```

### Gradio U

```python
from iointel import Agent, register_tool

@register_tool
def get_weather(city: str) -> dict:
    return {"temp": 72, "condition": "Sunny"}

agent = Agent(
    name="GradioSolar",
    instructions="You are a helpful assistant.",
    model="openai/gpt-oss-120b",
    api_key=api_key,
    base_url="https://api.intelligence.io.solutions/api/v1",
    tools=[get_weather],
    show_tool_calls=True,
)

await agent.launch_chat_ui(interface_title="Iointel Gradio Solar", share=False)


# Or, for more control across different agents:
# from iointel.src.ui.io_gradio_ui import IOGradioUI
# ui = IOGradioUI(agent, interface_title="Iointel GradioSolar")
# await ui.launch(share=False)
```
