# Ionet Cloud Virtual Machine

Ionet's virtual machines are similar to general AWS ECS servers, etc., and are the most commonly used type of server rental method.

For large model applications, they come pre-installed with frameworks like PyTorch and CUDA, and can directly run most machine learning code.

The specific rental process is similar to AWS and other cloud platforms. Here we mainly explain some key points.

The virtual machine service is:
https://cloud.io.net/cloud/virtual-machines

<img src='img/cloud2.png' width='600' />

Click the "deploy" button to start deployment.

## Step 1: Network Speed Confirmation

Select default.

<img src='img/netspeed.png' width='600' />

## Step 2: Choose Server Type

For learning and testing, generally choose 4090.

Note that you need to click on the graphics card type, then select the server's geographical location:

<img src='img/card.png' width='600' />

## Step 3: Determine Image and Add SSH Key

This step is mainly to determine which image to use. Generally, choose the second "Data science..." option, which comes with Jupyter service.

<img src='img/ssh.png' width='600' />

Software environment as follows:
- OS: Ubuntu 24.04.1 LTS (Noble Numbat)
- CUDA: 12.1
- RAPIDS: 25.6.0
- Python: 3.12
- Conda: 25.7.0

For SSH key, simply add your PC's public key content:

<img src='img/key.png' width='600' />


## Step 4: Add Additional Network Ports

Those who have used AWS and other servers know this - it's about opening additional network ports. By default, usually only common ports like 22 are available.

For Jupyter, the port is 8888, which needs to be specifically added, otherwise it cannot be accessed.

Also note that after the server is created, you cannot add ports anymore - they can only be added during creation.

<img src='img/port.png' width='600' />

## Login and Usage

After deployment is complete, the cluster page is displayed as follows:

<img src='img/cluster.png' width='600' />

Use the command shown in the SSH access to login. Windows generally requires PowerShell terminal, while Mac and Linux can use regular terminals.

The command format is generally as follows:

```
ssh -p 28282 ionet@52.205.25.210
```

After login, you can see the following prompt:

<img src='img/login.png' width='600' />



## Start Jupyter

First, start Jupyter by running the Jupyter startup script:
```
launch-jupyter
```

<img src='img/jupyter.png' width='600' />

During startup, you can see the login token in the startup logs, which looks like this:

http://127.0.0.1:8888/lab?token=1981227a4100db27f884280cb7494fdf8952481d0313a379

The parameter after token: 1981227a4100db27f884280cb7494fdf8952481d0313a379 is the login token, which is similar to general Jupyter usage.


## Using Jupyter

First, determine the external network address of Jupyter. This address is the same as the SSH login address. For example, if the SSH address is 52.205.25.210, the Jupyter login address is also 52.205.25.210.

Then for the port, Jupyter's service port is 8888, which we added in the network service, but note that this is the internal port.

For security reasons, ionet will map an external port. Specifically, click on network service and select the Jupyter service to see:

<img src='img/network.png' width='600' />

Here, port 8888 corresponds to external port 31814.

Then the Jupyter access address is:
http://52.205.25.210:31814

<img src='img/jplog.png' width='600' />

The token to fill in is the one obtained in the previous step: 1981227a4100db27f884280cb7494fdf8952481d0313a379

After filling it in, click login:

<img src='img/jp.png' width='600' />

Also note that Jupyter's web login page has WebSocket services, so it's best not to use VPN and proxy when accessing, otherwise it will be very slow or even inaccessible.

Note this. After login, using VPN is not a big problem.

## Environment Testing

You can use the following code to test if the environment is working properly:

```
import torch

print(f"PyTorch version: {torch.__version__}")

# Check if CUDA is detected by PyTorch
is_available = torch.cuda.is_available()

print(f"Is CUDA available? {is_available}")

# Get the current default GPU index
current_device_index = torch.cuda.current_device()
print(f"Current GPU index: {current_device_index}")

# Get the current GPU name
current_device_name = torch.cuda.get_device_name(current_device_index)
print(f"Current GPU name: {current_device_name}")
```
