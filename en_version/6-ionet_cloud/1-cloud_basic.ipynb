# Ionet Cloud Service Introduction

## io.net Cloud Core Services

`io.net Cloud` provides a series of flexible deployment options designed to meet different AI computing needs from individual developers to enterprise-level users. Its main services include:

1. **Container Engine**
   * **Introduction**: This is a service designed for modern developers, allowing users to deploy their Docker containers directly to `io.net`'s high-performance GPU clusters. It greatly simplifies environment configuration and is the most convenient way to quickly start model training and AI applications.

2. **Virtual Machine**
   * **Introduction**: Provides a complete virtual machine environment pre-installed with mainstream AI frameworks (such as PyTorch, TensorFlow), just like using AWS EC2. It's perfect for users who need persistent, interactive development environments, such as running Jupyter Notebook for data exploration and model debugging.

3. **Ray Cluster**
   * **Introduction**: Tailored for large-scale distributed Python and AI application developers. Users can deploy a decentralized Ray cluster with one click, easily scaling complex distributed training or hyperparameter tuning tasks to `io.net`'s global GPU network.

4. **Bare Metal on Demand**
   * **Introduction**: For expert users who need ultimate performance and hardware control. This service provides direct access to physical servers (bare metal), eliminating virtualization layer performance overhead, suitable for demanding tasks like HPC (High Performance Computing).

5. **Kubernetes/Not Yet Available (Kubernetes Cluster)**
   * **Introduction**: An enterprise-oriented service that allows `io.net`'s decentralized GPU resources to be used as compute nodes, seamlessly integrated into existing Kubernetes workflows. This enables enterprises to leverage `io.net`'s low-cost computing power to elastically scale their cloud-native AI applications.

**In summary**, `io.net Cloud` provides users with a **low-cost, large-scale, and highly flexible** decentralized AI computing platform by embracing industry standards like Docker, virtual machines, Ray, and Kubernetes.

For users, the services can be roughly divided into two categories:
1. Professional services. Including Ray clusters and bare metal on demand, these two are mainly for training large-scale models, requiring relatively professional large model knowledge, such as Ray learning, etc. Generally, multiple machines are used to form clusters, and then training services are run.
2. Learning and development. Including container engine and virtual machines, mainly single machine/single GPU, similar to services provided by autodl, vast.ai, etc., suitable for learning and development use, and basic fine-tuning is also possible.

## Service Pricing

Ionet's cloud service is essentially a GPU rental intermediary service, which requires payment to GPU providers, so this part is charged, but the cost is relatively cheap, especially compared to AWS, etc.

Additionally, because it's essentially a C2C rental service, you can also rent gaming cards like 4090, which doesn't violate NVIDIA's regulations.

For payment, it currently mainly accepts io tokens or USDT token payments. If needed, you can purchase them from exchanges.

It also supports general Visa, Master and other credit card payments.

The rental prices are generally low, but will have slight adjustments based on supply and demand. Machine quantities and prices can be viewed on the page:

https://explorer.io.net/explorer/home

For a startup company, being able to provide thousands of cards at a cheap price is already very good.
