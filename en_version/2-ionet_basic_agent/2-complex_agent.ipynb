## Text Classification Agent

Can you briefly explain what "text classification" is?

We can imagine it as a **librarian with superpowers**.

When a huge pile of disorganized books is delivered to a library, this librarian can quickly scan the content of each book and accurately place it on the correct shelf—such as "History," "Science," "Fiction," and so on.

---

### What is Text Classification?

**Text Classification** is one of the most fundamental and important tasks in Natural Language Processing (NLP). Its goal is to enable computers to **automatically assign one or more pre-defined "categories" or "labels" to a piece of text**.

Simply put, it allows machines to "understand" a piece of text and tell us, "What is this text about?"

### A Simple Example

Suppose we want the computer to automatically determine the sentiment of movie reviews. We can define three categories in advance: `positive`, `negative`, and `neutral`.

* **Input Text 1:**
  > "This movie is fantastic! The acting is excellent and the plot is very engaging!"

* **Computer's Output Label:**
  > `positive`

* **Input Text 2:**
  > "A complete waste of time. The story makes no sense—don't recommend watching."

* **Computer's Output Label:**
  > `negative`

This process is a typical example of text classification.

---

### How Does a Computer "Learn" to Classify?

Computers don't naturally know how to classify; they need a "learning" process, typically involving two steps:

1.  **Training**:
    We first prepare tens of thousands of articles that have already been manually labeled (this is like giving the computer a "workbook with answer keys"). Then, we let an AI model (such as a modern large language model) "learn" from this data. The model discovers patterns on its own—for example, articles containing words like "fantastic" or "recommend" are likely `positive`, while those with words like "disappointing" or "waste of time" are likely `negative`.

2.  **Prediction (Inference)**:
    Once the model has been trained, we can give it a completely new, unlabeled article that it has never seen before. It will use the knowledge and patterns learned during training to predict which category the article most likely belongs to.

---

### Why is Text Classification Important? (Key Applications)

Text classification is everywhere—it's at the core of many intelligent applications:

* **Spam Detection**
  * This is the classic text classification use case. Your email system automatically sorts incoming messages into "Inbox" or "Spam" folders.

* **Sentiment Analysis**
  * Analyzing social media posts (e.g., on Twitter or Weibo) or product reviews on e-commerce sites to determine whether users' opinions about a product, brand, or event are positive, negative, or neutral.

* **News Categorization**
  * Apps like Toutiao or Google News automatically classify news articles into channels such as "Sports," "Technology," "Finance," and "Entertainment."

* **Intelligent Customer Support and Ticketing Systems**
  * Automatically routing user support requests or complaint emails to the appropriate department based on content (e.g., "Technical Issue," "Billing Inquiry," "Shipping Question").

* **Content Moderation**
  * Automatically identifying harmful text content online—such as hate speech, violence, or spam ads—and taking appropriate action.

**In short, the core value of text classification is to transform massive amounts of unstructured text into well-organized, structured information**, enabling computers to process and utilize this data more efficiently. It is a crucial step toward information automation.

Before the era of large models, text classification typically required dedicated data labeling and model training. ## Named Entity Recognition Agent

We can imagine it as using a computer to **automatically highlight key information in a text with different colored fluorescent markers**.

---

### What is Named Entity Recognition?

**Named Entity Recognition** (NER) is a fundamental and core task in Natural Language Processing (NLP). Its goal is to **identify and classify "entities" of predefined types from a piece of text**.

Simply put, it involves extracting key pieces of information from a sentence—such as "**Who**", "**What**", "**When**", and "**Where**".

### A Simple Example

Consider the following sentence:

> **Apple Inc.**'s CEO, **Tim Cook**, released a new iPhone on **September 10, 2024**, in **California**'s **Cupertino**.

If a computer performs Named Entity Recognition on this sentence, it would highlight the key parts like this:

* `Apple Inc.` → **Organization**
* `Tim Cook` → **Person**
* `September 10, 2024` → **Date**
* `California` → **Location**
* `Cupertino` → **Location**

The computer not only identifies these terms but also accurately assigns them to predefined categories (e.g., person, location, etc.).

### What Are the Common Entity Types?

There are many types of entities that NER can recognize. The most common ones include:

* **PER**: Person
* **ORG**: Organization (e.g., companies, government agencies)
* **LOC**: Location (e.g., countries, cities, streets)
* **GPE**: Geo-Political Entity (specifically referring to administrative regions like countries and cities)
* **TIME**: Time
* **DATE**: Date
* **MONEY**: Monetary values
* **PRODUCT**: Product names
* **EVENT**: Event names

### Why is Named Entity Recognition Important? (Key Applications)

NER serves as the foundation for many advanced AI applications. It transforms unstructured text into structured data and has widespread uses:

1.  **Information Extraction and Intelligent Search**
    * When you ask a search engine, "**Who is the CEO of Baidu?**", NER first identifies "Baidu" as an **Organization**, enabling the system to look up the "CEO" information associated with "Baidu" in its knowledge base.

2.  **Content Recommendation and Categorization**
    * News websites use NER to identify people, locations, and companies mentioned in articles, helping determine whether the article belongs to technology, politics, or sports, and then recommend similar content accordingly.

3.  **Intelligent Customer Service**
    * Automatically extract entities such as **product names**, **order numbers**, **dates**, and **amounts** from customer complaint emails to quickly create service tickets and improve efficiency.

4.  **Resume Parsing**
    * Automatically extract **name**, **university attended**, **previous employers**, and **technical skills** from job applicants' resumes and store them in a talent database.

5.  **Sentiment and Public Opinion Analysis**
    * Analyze social media discussions about a specific **brand** or **product** to quickly understand public sentiment and key concerns.

**In summary, Named Entity Recognition is the first step in enabling computers to understand critical information in text**, serving as a vital bridge between human language and structured data.

Of course, in the era of large language models, NER no longer requires complex training procedures—just a well-crafted prompt is sufficient.

## Custom Agent

Customize agents through prompts. Prompts can customize various tasks, including all the previous types of tasks.

```python
from iointel import (
    Agent,
    Workflow
)

import os
import asyncio

api_key= "io-v2-eyJhbGciOiJSUzI1NiIsInR5cCI6IkpXVCJ9.eyJvd25lciI6IjAzNzRhMjJjLTVmZTctNDBhNS04OGU4LTEwMjk3MGViMjEyNiIsImV4cCI6NDkwNTc2MTYyM30.F2HzmnMxRaHpRh2kfPRSgp7JHqtVb5BUmZD4kz1TxY28Usvyh18shHpNpvmncoE8S-IvZH3qlEEScNtO9j-5Dw"

text = """The global electric vehicle (EV) market is witnessing exponential growth, driven by advancements in battery technology and government incentives. Major automakers, including Tesla, Ford, and Volkswagen, are investing heavily in EV production. However, challenges such as charging infrastructure and raw material shortages for lithium-ion batteries remain key concerns. Industry experts predict that by 2030, EVs will account for over 50% of new car sales worldwide, marking a significant shift in the automotive landscape."""

agent = Agent(
    name="Super Agent",
    instructions="You are an assistant specialized in doing anything.",
    model="meta-llama/Llama-3.3-70B-Instruct",
    api_key=api_key,
    base_url="https://api.intelligence.io.solutions/api/v1"
)

workflow = Workflow(objective=text, client_mode=False)

async def run_workflow():
    results = (await workflow.custom(name="custom-task", objective="continue summarizing",instructions="Focus on previous history",agents=[agent],).run_tasks())["results"]
    return results

results = await run_workflow()
print(results)
```
