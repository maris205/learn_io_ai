## Basic Large Language Model Calling

ionet's large model service calls are fully compatible with OpenAI's SDK. The key points are as follows:

### 1. Install the OpenAI Python Library

First, ensure that your Python environment has the official OpenAI library installed. You can install it using the `pip` package manager:

<pre style="background: none"><code class="language-bash" data-language="bash" identifier="d7943b1a8a7847d48b40af3f38ce25f0-0" index="0" total="6">pip install --upgrade openai</code></pre>

The `--upgrade` parameter ensures you are installing the latest version, which is important because OpenAI's API and library are updated periodically.

### 2. Obtain and Set the API Key

To use OpenAI's service, you must have an API key. Third-party services usually have a similar key as well.

* **Obtain the key**: Log in to your OpenAI account and create a new key on the "API keys" page. Note that the key will only be displayed in full once, so be sure to store it securely and avoid disclosing it to others.

* **Set the key**: The recommended and most secure way to use the key in code is to set it as an environment variable.

  * On **Linux or macOS**, run the following command in your terminal:
    <pre style="background: none"><code class="language-bash" data-language="bash" identifier="d7943b1a8a7847d48b40af3f38ce25f0-1" index="1" total="6">export OPENAI_API_KEY='your-key'</code></pre>
  * On **Windows**, run:
    <pre style="background: none"><code class="language-bash" data-language="bash" identifier="d7943b1a8a7847d48b40af3f38ce25f0-2" index="2" total="6">set OPENAI_API_KEY=your-key</code></pre>
    To make this setting permanent, you need to add it to your system's environment variables.

When you initialize the OpenAI client in your code, it will automatically read the environment variable named `OPENAI_API_KEY`, eliminating the need to hardcode the key in your code, which is much more secure.

Additionally, most large model services compatible with the OpenAI SDK generally require setting a `base_url` to indicate the specific server address of the service.

### 3. Initialize the Client

In your Python code, the first step is to import the `openai` library and create a client instance.

<pre style="background: none"><code class="language-python" data-language="python" identifier="d7943b1a8a7847d48b40af3f38ce25f0-3" index="3" total="6">from openai import OpenAI

# If you've already set the environment variable, the client will automatically read the key
client = OpenAI()

# If you haven't set the environment variable (not recommended), you can specify it directly in the code
# client = OpenAI(api_key="your-key")</code></pre>

### 4. Call the Large Language Model (Core Step)

The most common usage is the "Chat Completions" interface. This interface is very flexible and can be used not only for multi-turn conversations but also for single text generation tasks.

Below is a basic example of making a call to ask the model a question:

<pre style="background: none"><code class="language-python" data-language="python" identifier="d7943b1a8a7847d48b40af3f38ce25f0-4" index="4" total="6">import os
from openai import OpenAI

# Initialize the client (automatically reads the OPENAI_API_KEY environment variable)
client = OpenAI()

try:
    # Make the API request
    response = client.chat.completions.create(
        model="gpt-4o",  # Specify the model to use, e.g., gpt-4o, gpt-4-turbo, gpt-3.5-turbo
        messages=[
            {"role": "system", "content": "You are a helpful AI assistant."},  # System message to set the AI's role and behavior
            {"role": "user", "content": "Please briefly introduce what's fun at Universal Studios Singapore?"}  # User's question
        ]
    )

    # Extract and print the model's response
    ai_response = response.choices[0].message.content
    print(ai_response)

except Exception as e:
    print(f"Error calling API: {e}")</code></pre>

#### Key Points of the Code:

* **`model`**: Specifies the model you want to use. `gpt-4o` is one of the latest and most comprehensive models, offering great value. You can also choose other models.
* **`messages`**: This is the core of the request, a list containing one or more dictionaries.
  * **`role`**: The role, mainly three types:
    * `system`: The system role, used to set the overall behavior and context for the model, e.g., "You are a professional translator."
    * `user`: The user role, representing the end user's question or instruction.
    * `assistant`: The assistant role, representing the model's previous replies. In multi-turn conversations, you need to include the entire history (user questions and AI responses) in the `messages` list so the model can understand the context.
  * **`content`**: The actual content of the message.
* **`response`**: The API returns a complex object. Typically, we care most about `response.choices[0].message.content`, which contains the text response generated by the model.

### 5. Advanced Usage: Controlling Output

You can also include additional parameters in the `create()` method to finely control the model's output:

* **`temperature`** (temperature, 0 to 2, default 1): Controls the randomness of the output. A higher value (e.g., 1.2) makes the output more random and creative; a lower value (e.g., 0.2) makes the output more deterministic and conservative.
* **`max_tokens`** (maximum tokens): Sets the maximum length of the content generated by the model. This helps control costs and response time. Note that a single word is usually composed of multiple tokens.
* **`stream`** (streaming): When set to `True`, the API returns results word by word like a typewriter, rather than waiting for the entire response to be generated before returning it all at once. This is very useful for achieving real-time chat effects.

#### Streaming Example:

<pre style="background: none"><code class="language-python" data-language="python" identifier="d7943b1a8a7847d48b40af3f38ce25f0-5" index="5" total="6"># ... (previous code is the same) ...

response_stream = client.chat.completions.create(
    model="gpt-4o",
    messages=[
        {"role": "user", "content": "Write me a short story about space travel"}
    ],
    stream=True  # Enable streaming
)

print("AI is generating:")
for chunk in response_stream:
    # Check if there is content and print it
    if chunk.choices[0].delta.content is not None:
        print(chunk.choices[0].delta.content, end="", flush=True)

print() # New line</code></pre>

### Summary

The basic process of using the OpenAI SDK is: **Install -> Obtain Key -> Initialize Client -> Construct `messages` List -> Call `chat.completions.create` Method -> Parse the Response**.

### Multi-turn Conversation

```python
import openai

# --- Initialize client ---
# Please replace api_key with your own key
client = openai.OpenAI(
    api_key="io-v2-eyJhbGciOiJSUzI1NiIsInR5cCI6IkpXVCJ9.eyJvd25lciI6IjAzNzRhMjJjLTVmZTctNDBhNS04OGU4LTEwMjk3MGViMjEyNiIsImV4cCI6NDkwNTc2MTYyM30.F2HzmnMxRaHpRh2kfPRSgp7JHqtVb5BUmZD4kz1TxY28Usvyh18shHpNpvmncoE8S-IvZH3qlEEScNtO9j-5Dw",
    base_url="https://api.intelligence.io.solutions/api/v1/",
)

# --- Conversation history ---
# First add "system" message to set a role for AI.
conversation_history = [
    {"role": "system", "content": "You are a helpful AI assistant."},
]

print("--- Starting a two-turn conversation example ---")

# --- First turn conversation ---
# 1. Define the first user question
first_user_message = "Hello, please briefly introduce yourself."
print(f"You: {first_user_message}")

# 2. Add user question to history
conversation_history.append({"role": "user", "content": first_user_message})

# 3. Send request
first_response = client.chat.completions.create(
    model="meta-llama/Llama-3.3-70B-Instruct",
    messages=conversation_history,
    temperature=0.7,
    stream=False,
    max_completion_tokens=150
)

# 4. Get and print AI's response
first_assistant_message = first_response.choices[0].message.content
print(f"Assistant: {first_assistant_message}\n")

# 5. Add AI's response to history for next turn
conversation_history.append({"role": "assistant", "content": first_assistant_message})


# --- Second turn conversation ---
# 1. Define second user question (this question is based on previous conversation)
second_user_message = "Great. What can you help me with?"
print(f"You: {second_user_message}")

# 2. Add second question to history
conversation_history.append({"role": "user", "content": second_user_message})

# 3. Send request again with all history
second_response = client.chat.completions.create(
    model="meta-llama/Llama-3.3-70B-Instruct",
    messages=conversation_history, # Key point: here we pass the complete history including first turn conversation
    temperature=0.7,
    stream=False,
    max_completion_tokens=150
)

# 4. Get and print final response
second_assistant_message = second_response.choices[0].message.content
print(f"Assistant: {second_assistant_message}")

```

### Streaming Call

```python
import openai

# --- Initialize client ---
# Please replace api_key with your own key
client = openai.OpenAI(
    api_key="io-v2-eyJhbGciOiJSUzI1NiIsInR5cCI6IkpXVCJ9.eyJvd25lciI6IjAzNzRhMjJjLTVmZTctNDBhNS04OGU4LTEwMjk3MGViMjEyNiIsImV4cCI6NDkwNTc2MTYyM30.F2HzmnMxRaHpRh2kfPRSgp7JHqtVb5BUmZD4kz1TxY28Usvyh18shHpNpvmncoE8S-IvZH3qlEEScNtO9j-5Dw",
    base_url="https://api.intelligence.io.solutions/api/v1/",
)

# --- Initiate streaming request ---
# Key point 1: Set stream to True
response_stream = client.chat.completions.create(
    model="meta-llama/Llama-3.3-70B-Instruct",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        # To make the streaming effect more obvious, use a question that requires a longer answer
        {"role": "user", "content": "Tell me a short story about a robot who dreams of stars."},
    ],
    temperature=0.7,
    stream=True,  # Set to True to enable streaming
    max_completion_tokens=150
)

# --- Process and print data stream ---
print("Assistant: ", end="")

# Key point 2: Use for loop to iterate through returned streaming object
for chunk in response_stream:
    # Check if each chunk has content and ensure choices list is not empty
    if chunk.choices and chunk.choices[0].delta.content is not None:
        # Use end="" and flush=True to achieve typewriter effect
        print(chunk.choices[0].delta.content, end="", flush=True)

# After all content is printed, output a newline
print()

```

## Multimodal Large Model Basics

Multimodal large models generally take image + prompt as input, where the prompt is a question about the image.

Basic calling is as follows:

```python
import base64
from openai import OpenAI

# Configure client
client = OpenAI(
    api_key="io-v2-eyJhbGciOiJSUzI1NiIsInR5cCI6IkpXVCJ9.eyJvd25lciI6IjAzNzRhMjJjLTVmZTctNDBhNS04OGU4LTEwMjk3MGViMjEyNiIsImV4cCI6NDkwNTc2MTYyM30.F2HzmnMxRaHpRh2kfPRSgp7JHqtVb5BUmZD4kz1TxY28Usvyh18shHpNpvmncoE8S-IvZH3qlEEScNtO9j-5Dw",
    base_url="https://api.intelligence.io.solutions/api/v1/"
)

# Image file path
image_path = "img/tello.jpg"

# Define local image encoding function
def encode_image(image_path):
    with open(image_path, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode("utf-8")

local_image_path = "img/tello.jpg"

# Image input:
response = client.chat.completions.create(
    model="Qwen/Qwen2.5-VL-32B-Instruct",
    messages=[
        {
            "role": "user",
            "content": [
                {"type": "text", "text": "What's in the image?"},
                {
                    "type": "image_url",
                    "image_url": {
                        # "url": "https://yt-shanghai.tos-cn-shanghai.volces.com/tello.jpg"
                        # use Base64 encode local image，attention image/png image/jpg type
                        "url": f"data:image/jpg;base64,{encode_image(local_image_path)}"
                    }
                },
            ],
        }
    ],
)


#print(response.choices[0].message.content)
print(response.choices[0])
```
